```python
import numpy as np
import matplotlib.pyplot as plt

L = np.pi

# Define functions
def f1(x):
    return np.where(x > 0, 1, -1)

def f2(x):
    return x

def f3(x):
    return np.where(x > 0, 1, 0)

# Fourier coefficients
def an(f, n):
    x = np.linspace(-L, L, 1000)
    return (1/L) * np.trapz(f(x) * np.cos(n*np.pi*x/L), x)

def bn(f, n):
    x = np.linspace(-L, L, 1000)
    return (1/L) * np.trapz(f(x) * np.sin(n*np.pi*x/L), x)

def a0(f):
    x = np.linspace(-L, L, 1000)
    return (1/L) * np.trapz(f(x), x)

# Fourier series approximation
def fourier_series(f, x, N):
    result = a0(f)/2
    for n in range(1, N+1):
        result += an(f, n)*np.cos(n*np.pi*x/L) + bn(f, n)*np.sin(n*np.pi*x/L)
    return result
def plot_function(f, title):
    x = np.linspace(-6*np.pi, 6*np.pi, 2000)

    plt.figure(figsize=(10,5))
    plt.plot(x, f((x+np.pi)%(2*np.pi)-np.pi), label="Original")

    for N in [5, 20, 50]:
        plt.plot(x, fourier_series(f, x, N), label=f"N={N}")

    # 🔥 X-axis in terms of π
    ticks = np.arange(-6*np.pi, 7*np.pi, np.pi)
    labels = []

    for t in ticks:
        k = int(round(t/np.pi))
        if k == 0:
            labels.append("0")
        elif k == 1:
            labels.append("π")
        elif k == -1:
            labels.append("-π")
        else:
            labels.append(f"{k}π")

    plt.xticks(ticks, labels)

    plt.title(title)
    plt.legend()
    plt.grid()
    plt.show()

# Run for all
plot_function(f1, "Square Wave")
plot_function(f2, "Sawtooth")
plot_function(f3, "Step Function")
```

```python
def plot_gibbs(f, title):
    x = np.linspace(-2*np.pi, 2*np.pi, 2000)

    plt.figure(figsize=(12,5))

    # -------- Full view --------
    plt.subplot(1,2,1)
    plt.plot(x, f((x+np.pi)%(2*np.pi)-np.pi), 'k', label="Original")

    for N in [5, 20, 50]:
        plt.plot(x, fourier_series(f, x, N), label=f"N={N}")

    plt.title("Full View")
    plt.grid()
    plt.legend()

    # -------- Zoom near discontinuity --------
    plt.subplot(1,2,2)

    zoom_x = np.linspace(-1, 1, 1000)  # zoom near 0
    plt.plot(zoom_x, f((zoom_x+np.pi)%(2*np.pi)-np.pi), 'k', label="Original")

    for N in [5, 20, 50]:
        plt.plot(zoom_x, fourier_series(f, zoom_x, N), label=f"N={N}")

    # 🔥 Highlight discontinuity line
    plt.axvline(0, linestyle='--')

    plt.title("Zoom near x = 0 (Gibbs Phenomenon)")
    plt.grid()
    plt.legend()

    plt.show()
plot_gibbs(f1, "Square Wave")
```
